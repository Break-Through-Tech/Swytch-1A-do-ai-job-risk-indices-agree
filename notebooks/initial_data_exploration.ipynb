```python
import os
import pandas as pd

# Path to the data folder (relative to where this script/notebook runs)
data_folder = '../data'

# Confirm it exists before looping — fails fast with a clear message instead of a cryptic error per file
if not os.path.isdir(data_folder):
    raise FileNotFoundError(f"'{data_folder}' not found relative to {os.getcwd()} — check your path")

files = os.listdir(data_folder)

count = 0

for file in files:
    file_path = os.path.join(data_folder, file)
    try:
        if file.endswith('.csv'):
            df = pd.read_csv(file_path)
            count += 1
        elif file.endswith('.xlsx'):
            df = pd.read_excel(file_path)
            count += 1
        else:
            continue
        print(f"Able to read {file}:")
        # print(df.head())
        print("\n")
    except Exception as e:
        print(f"Could not read {file}: {e}")
    
    print(count)
```
