## Task 7 Summary - Neisha

### BLS Employment Projections Table 1.2

The project BLS workbook contains Table 1.2 for the 2025–35 projection period.

### BLS data preparation
- Raw Table 1.2 rows: 1,117
- Line-item occupations: 831
- Summary rows: 281
- Footer/note rows: 5
- Unique line-item occupation codes: 831
- All 831 line-item codes follow the `##-####` occupation-code format.
- `Employment change, percent, 2025–35` was numeric with no missing or unavailable (`—`) values among the line-item occupations.

### Join to merged_v0
- `merged_v0` occupations: 798
- Exact BLS matches: 772
- Unmatched occupations: 26
- Exact match rate: 96.74%
- Rows after left join: 798

A left join was used so that every occupation already in `merged_v0` was preserved even when an exact BLS Employment Matrix code was unavailable.

### Unmatched occupations
The 26 unmatched SOC codes were absent from the complete BLS Table 1.2, not only from the filtered line-item data.

Investigation showed that many unmatched detailed occupations are represented in the BLS Employment Matrix by broader combined occupation codes. Examples include:

- `13-1021`, `13-1022`, and `13-1023` represented by BLS `13-1020`, Buyers and purchasing agents.
- `29-2011` and `29-2012` represented by BLS `29-2010`, Clinical laboratory technologists and technicians.
- `31-1121` and `31-1122` represented by BLS `31-1120`, Home health and personal care aides.
- `39-7011` and `39-7012` represented by BLS `39-7010`, Tour and travel guides.

The broader BLS projection was not automatically copied to the detailed occupations because that would require a methodological decision about assigning one aggregated projection to multiple SOC occupations.
