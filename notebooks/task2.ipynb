```python
import pandas as pd

df = pd.read_csv('../data/datasets/occ_level.csv') 

#Look into the first 20 rows of the dataset
df.head(20)
```

```python
#Data exploration to ensure we have no null values
print(df.isnull().sum())

df.info()
```

```python
# Glance at O*NET-SOC Code that have values after the decimal point
onet = df['O*NET-SOC Code']

special_row = []
number_to = []

for num in onet:
    if int(num[8:10]) > 0:
        detail_row = df[df['O*NET-SOC Code'] == num]
        number_to.append(num)
        special_row.append(detail_row)

detail_rows = pd.concat(special_row)  

print("Number of rows with O*NET-SOC Code having values after the decimal point:", len(detail_rows))
detail_rows
```

```python
#Get list of column names which we would operate on and convert to numeric values

list(df.columns)
operate_on = df.drop(columns=['O*NET-SOC Code', 'Title']).columns.tolist()
print("Columns to operate on:", operate_on)

df[operate_on] = df[operate_on].apply(pd.to_numeric, errors='coerce')
df.dtypes
```

```python
# Create a title_lookup so we can pair the correct title and O*NET-SOC (Base) Code with the correct row in the dataframe
title_lookup = df[df['O*NET-SOC Code'].str.endswith('.00')][['O*NET-SOC Code', 'Title']]
title_lookup
```

```python
# Make sure soc_6digit columns exist --> removes the decimal point and everything after it, leaving only the 6-digit family code
df['soc_6digit'] = df['O*NET-SOC Code'].str.split('.').str[0]

# Group all rows (base + detail codes) by their 6-digit family, and calculate the mean of each column in operate_on
df_processed = df.groupby('soc_6digit')[operate_on].mean().reset_index()



# Add soc_6digit to title_lookup so it can be merged with df_processed
title_lookup['soc_6digit'] = df['soc_6digit']

# Merge titles into df_processed by soc_6digit
df_processed = df_processed.merge(title_lookup[['soc_6digit', 'Title']], on='soc_6digit', how='left')

# Move Title to right after soc_6digit
df_processed.insert(1, 'Title', df_processed.pop('Title'))

print(f"Original rows: {len(df)}, After rollup: {len(df_processed)}")
    
```

```python
#Display the final dataset after rollup and merging titles
df_processed
```

# Note: SOC 8-digit → 6-digit Rollup — Reduction Count Discrepancy

## Finding

The raw dataset contains **923** occupations at the 8-digit O*NET-SOC level. After
rolling up to 6-digit SOC codes, the final dataset contains **798** unique codes —
a reduction of **125** rows.

At first glance this doesn't match expectations: there are **149** detail-level
rows in the raw data (codes with a suffix other than `.00`). If every detail row
simply merged into an existing base (`.00`) row, the reduction should also be 149,
not 125.

## Why the numbers differ

The gap (149 − 125 = **24**) is explained by **24 six-digit SOC families that have
no `.00` base row at all** — they exist *only* as detail-level codes. For example,
`29-1299` and `19-3099` have multiple `.XX` detail codes but no general `.00` entry.

For these 24 families, the detail rows aren't "extra" rows stacked on top of a base
— they *are* the entire group. Collapsing them into one 6-digit row doesn't reduce
the row count the same way it does for a family that already had a base row, which
is why the total reduction comes out lower than a naive count of detail rows would
suggest.



```python
#Verification of the rollup and merging of titles

df['suffix'] = df['O*NET-SOC Code'].str.split('.').str[1].astype(int)

group_sizes = df.groupby('soc_6digit').size()

has_base = set(df.loc[df['suffix'] == 0, 'soc_6digit'].unique())
all_groups = set(group_sizes.index)
no_base_groups = all_groups - has_base

detail_count = (df['suffix'] != 0).sum()
excess = (group_sizes - 1).sum()

print(f"Total rows: {len(df)}")
print(f"Detail rows (suffix != .00): {detail_count}")
print(f"Unique 6-digit groups: {len(group_sizes)}")
print(f"Reduction: {len(df) - len(group_sizes)}")
print(f"6-digit codes with NO .00 base row: {len(no_base_groups)}")
print(sorted(no_base_groups))

print("Reconciliation check — detail_count - excess should equal no_base count:",
      detail_count - excess, "vs", len(no_base_groups))
```

# Methods Note: SOC 8-digit → 6-digit Rollup

## Rollup Rule

Exposure scores were rolled up from the 8-digit O*NET-SOC level to the 6-digit
SOC level using a **simple (unweighted) mean** across all detail rows within
each 6-digit family.

## Rationale

Mean was chosen as a fast, defensible starting point to keep the pipeline
moving — not as a deliberated final answer. Alternatives (e.g. employment-weighted
mean using OES employment counts, or min/max bounding for conservative vs.
liberal exposure estimates) were **not evaluated in depth** at this stage.
That comparison is better suited to a team-level discussion during the
crosswalk stage, since the choice affects every downstream index comparison.
The current mean-based approach is easy to swap out later if the team decides
on a different weighting scheme.

## Known Limitation: Missing Titles

24 of the 798 six-digit SOC codes have no `Title` populated. This is because
these families have no `.00` base row to pull a title from — they exist only
as detail-level O*NET-SOC codes.

This has been **left as a documented gap rather than resolved ad hoc**:
- Resolving it properly requires a crosswalk against the official BLS SOC
  structure/title list, which is outside this task's scope.
- Manually assigning titles from memory or guesswork risked introducing
  incorrect labels and could conflict with the separate crosswalk effort
  another team member may own.

## Verification

The 923 → 798 row reduction was checked against the underlying data:
149 detail-level rows (non-`.00` suffix) exist in the raw dataset, but the
net reduction is 125, not 149. This gap (24) is fully explained by the 24
SOC families noted above that have no base row — their detail rows collapse
into a single group without producing the same row reduction as families
that already had a base row. This verification is considered solid and
stated here as fact, not an open question.
